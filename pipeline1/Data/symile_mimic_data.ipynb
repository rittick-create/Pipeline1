```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display
ROOT = next(p / "pipeline1" for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "pipeline1/Stage1_MAE.py").is_file())
sys.path.insert(0, str(ROOT))
```

```python
from Data.create_datasets import SymileMIMICDataset
DATA_ROOT = ROOT.parent / "symile-mimic/symile-mimic-aws"
train = SymileMIMICDataset(DATA_ROOT, "train", "distillation")
val = SymileMIMICDataset(DATA_ROOT, "val", "distillation")
print("Training admissions", len(train), "Validation admissions", len(val))
print("Patient overlap", bool(set(train.subject_ids) & set(val.subject_ids)))
for split, dataset in [("train", train), ("val", val)]:
    print(split)
    sample = dataset[0]
    display(pd.DataFrame([{"input": key, "shape": tuple(value.shape),
                           "dtype": str(value.dtype)} for key, value in sample.items()]))
```
